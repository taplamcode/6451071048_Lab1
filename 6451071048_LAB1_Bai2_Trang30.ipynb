```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('titanic.csv')

# Biểu đồ cột thể hiện số lượng hành khách sống sót theo Giới tính
survived_by_sex = df.groupby('sex')['survived'].value_counts().unstack()
survived_by_sex.plot(kind='bar', color=['#e74c3c', '#2ecc71'], figsize=(8, 5))

plt.title("Số lượng hành khách Sống sót / Thiệt mạng theo Giới tính")
plt.xlabel("Giới tính")
plt.ylabel("Số người")
plt.legend(['Thiệt mạng (0)', 'Sống sót (1)'])
plt.xticks(rotation=0)
plt.show()
```

```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('titanic.csv')

# Biểu đồ đường thể hiện giá vé trung bình theo Độ tuổi
age_fare = df.groupby(pd.cut(df['age'], bins=10))['fare'].mean()

plt.figure(figsize=(9, 5))
age_fare.plot(kind='line', marker='o', color='purple')
plt.title("Giá vé trung bình theo từng khoảng tuổi trên tàu Titanic")
plt.xlabel("Khoảng độ tuổi")
plt.ylabel("Giá vé trung bình ($)")
plt.grid(True)
plt.show()
```

```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('titanic.csv')

# Biểu đồ phân tán Scatter Plot thể hiện mối liên hệ giữa Age và Fare
plt.figure(figsize=(8, 5))
plt.scatter(df['age'], df['fare'], alpha=0.5, c=df['survived'], cmap='coolwarm')
plt.title("Mối liên hệ giữa Độ tuổi và Giá vé (Màu đại diện cho Sống/Chết)")
plt.xlabel("Độ tuổi (Age)")
plt.ylabel("Giá vé (Fare)")
plt.colorbar(label='Survived (0 = No, 1 = Yes)')
plt.show()
```

```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('titanic.csv')

# Biểu đồ hộp Boxplot kiểm tra ngoại lệ giá vé theo hạng vé
df.boxplot(column='fare', by='pclass', figsize=(8, 5))
plt.title("Biểu đồ Boxplot phân bố Giá vé theo Hạng vé")
plt.suptitle("")
plt.xlabel("Hạng vé (Pclass)")
plt.ylabel("Giá vé (Fare)")
plt.show()
```

```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('titanic.csv')

# Ma trận tương quan (Heatmap) giữa các chỉ số định lượng
numeric_cols = df[['survived', 'pclass', 'age', 'sibsp', 'parch', 'fare']]
plt.figure(figsize=(8, 6))
sns.heatmap(numeric_cols.corr(), annot=True, cmap='YlGnBu', fmt=".2f")
plt.title("Ma trận tương quan Heatmap trên tập Titanic")
plt.show()
```
