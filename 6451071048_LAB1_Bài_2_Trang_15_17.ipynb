```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd

df = pd.read_csv('titanic.csv')

# Kiểm tra giá trị thiếu (Missing values) trên từng cột
missing_info = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({'Số lượng khuyết': missing_info, 'Tỷ lệ (%)': missing_percent})
print("Thống kê dữ liệu bị khuyết thiếu trên tập Titanic:")
print(missing_df[missing_df['Số lượng khuyết'] > 0])
```

```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd

df = pd.read_csv('titanic.csv')

# Xử lý missing values: Điền Mean cho 'age', Median cho 'fare', Mode cho 'embarked'
df['age'] = df['age'].fillna(df['age'].mean())
df['fare'] = df['fare'].fillna(df['fare'].median())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

print("Số lượng giá trị khuyết sau khi xử lý:")
print(df[['age', 'fare', 'embarked']].isnull().sum())
print("\nXem mẫu dữ liệu sau khi điền khuyết:")
print(df[['name', 'age', 'fare', 'embarked']].head())
```

```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd

df = pd.read_csv('titanic.csv')

# Lọc hành khách nữ thuộc vé hạng 1 (pclass = 1) có độ tuổi > 25
filtered_df = df[(df['sex'] == 'female') & (df['pclass'] == 1) & (df['age'] > 25)]

print(f"Số lượng hành khách thỏa điều kiện: {len(filtered_df)}")
print("\nDanh sách 5 hành khách đầu tiên thỏa điều kiện:")
print(filtered_df[['name', 'pclass', 'sex', 'age', 'survived']].head())
```

```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd

df = pd.read_csv('titanic.csv')

# Gom nhóm theo Hạng vé (pclass) và Giới tính (sex), tính trung bình độ tuổi và giá vé
grouped = df.groupby(['pclass', 'sex'])[['age', 'fare', 'survived']].mean()

print("Thống kê giá trị trung bình theo Hạng vé và Giới tính:")
print(grouped)
```
