```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv('titanic.csv')
df['fare'] = df['fare'].fillna(df['fare'].median())

# Chuẩn hóa Min-Max Scaling cho cột Fare (Giá vé)
scaler = MinMaxScaler()
df['fare_minmax'] = scaler.fit_transform(df[['fare']])

print("Mẫu dữ liệu trước và sau khi chuẩn hóa Min-Max Scaling [0, 1]:")
print(df[['fare', 'fare_minmax']].head())
```

```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('titanic.csv')
df['age'] = df['age'].fillna(df['age'].mean())

# Chuẩn hóa Z-Score (StandardScaler) cho cột Age (Tuổi)
scaler = StandardScaler()
df['age_zscore'] = scaler.fit_transform(df[['age']])

print("Mẫu dữ liệu trước và sau khi chuẩn hóa Z-Score (Mean=0, Std=1):")
print(df[['age', 'age_zscore']].head())
```

```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd

df = pd.read_csv('titanic.csv')
df['fare'] = df['fare'].fillna(df['fare'].median())

# Phát hiện và loại bỏ Outlier bằng phương pháp IQR trên cột 'fare'
Q1 = df['fare'].quantile(0.25)
Q3 = df['fare'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_clean = df[(df['fare'] >= lower_bound) & (df['fare'] <= upper_bound)]

print(f"Số lượng dòng ban đầu: {len(df)}")
print(f"Số lượng dòng sau khi lọc Outliers (cột fare): {len(df_clean)}")
```

```python
print("Họ tên: Trần Thanh Ngân - MSSV: 6451071048")
print("=" * 60)

import pandas as pd

df = pd.read_csv('titanic.csv')
df['age'] = df['age'].fillna(df['age'].mean())

# Rời rạc hóa (Binning) cột 'age' thành 3 nhóm tuổi
bins = [0, 18, 50, 100]
labels = ['Trẻ em', 'Người trưởng thành', 'Người cao tuổi']
df['nhom_tuoi'] = pd.cut(df['age'], bins=bins, labels=labels)

print("Phân bố số lượng theo nhóm tuổi:")
print(df['nhom_tuoi'].value_counts())
```
